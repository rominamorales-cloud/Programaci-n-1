## Excepciones

"El agua hierve a 100 grados bajo cero cuando hace calor."

Aunque esta oración es correcta desde el punto de vista gramatical, su significado está lleno de contradicciones. Algo similar ocurre en la programación: incluso cuando el código tiene la sintaxis correcta, puede generar errores lógicos durante la ejecución. A estos errores se les conoce como excepciones, y son detectados cuando el programa se está ejecutando, generando mensajes de error que pueden, en muchos casos, ser controlados y gestionados.

Ejecuta las siguientes líneas y observa que ocurre...

```python
10 * (1/0)
```

```python
4 + spam*3
```

```python
'2' + 2
```

La última línea del mensaje de error es crucial, ya que indica qué tipo de excepción ha ocurrido. Cada tipo de excepción tiene un propósito específico y se utiliza para señalar un tipo particular de error.

```python
#Desafío 67

#Ejecutar el programa
try:
    #Se pide dos números enteros al usuario
    num1 = int(input("Ingresa el primer número entero: "))
    num2 = int(input("Ingresa el segundo número entero: "))
    #División
    resultado = num1 / num2
    #Resultado
    print("El resultado es:", resultado)

#Error si se divide entre cero
except ZeroDivisionError:
    print("Error: no se puede dividir entre cero.")

#Error si se ingresan datos no numéricos
except ValueError:
    print("Error: debes ingresar números enteros.")
```

```python
#Desafío 68
#Se crea una lista con números y un texto
valores = [10, 5, 0, "hola"]

try:
    #Se recorre cada elemento de la lista
    for valor in valores:

        #Se convierte el valor a número decimal
        numero = float(valor)

        #Se divide 100 entre el número
        resultado = 100 / numero

        #Resultado
        print("Resultado:", resultado)

#Error de división por cero
except ZeroDivisionError:
    print("Error: no se puede dividir entre cero.")

#Errores de tipos de datos incompatibles
except TypeError:
    print("Error: tipo de dato incompatible.")

#Errores al convertir un texto a número
except ValueError:
    print("Error: el valor no es numérico.")
```

```python
#Desafío 69
import math

#Función que calcula el factorial
def factorial(numero):
    #Se comprueba que el número sea entero
    if not isinstance(numero, int):
        raise TypeError("El número debe ser entero.")

    #Se comprueba que el número no sea negativo
    if numero < 0:
        raise ValueError("El número no puede ser negativo.")

    #Se establece un límite para el ejercicio
    if numero > 1000:
        raise OverflowError("El número es demasiado grande.")

    #Se calcula y se devuelve el factorial
    return math.factorial(numero)

try:
    #Se pide un número entero
    numero = int(input("Ingresa un número entero: "))

    resultado = factorial(numero)
    print("El factorial es:", resultado)

#Errores de conversión o números negativos
except ValueError as e:
    print("Error:", e)
#Errores relacionados con el tipo de dato
except TypeError as e:
    print("Error:", e)
#Error si se supera el límite establecido
except OverflowError as e:
    print("Error:", e)
```

```python
#Desafío 70
#Se crea la excepción personalizada
class NegativeNumberError(Exception):
    pass

#Calcular la raíz cuadrada
try:
    #Se pide un número decimal al usuario
    numero = float(input("Ingresa un número: "))
    #Se comprueba si el número es negativo
    if numero < 0:
        raise NegativeNumberError(
            "No se permiten números negativos."
        )
    import math
    resultado = math.sqrt(numero)
    print("La raíz cuadrada es:", resultado)

#Errores de entrada no numérica
except ValueError:
    print("Error: debes ingresar un número válido.")
#Excepción personalizada
except NegativeNumberError as e:
    print("Error:", e)
```
